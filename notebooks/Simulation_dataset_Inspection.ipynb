```python
import os, h5py, numpy as np

from pathlib import Path
ROOT = Path.cwd().parent
SRC = ROOT / "data" / "test_BK_0d224_1d073_bak_surf1_pert.mat"
OUT = ROOT / "data" / "for_claude"
TAG = "BK_0d224_1d073"
os.makedirs(OUT, exist_ok=True)

with h5py.File(SRC, "r") as f:
    nt = int(np.array(f["time_len_min"]).ravel()[0])
    x = np.array(f["x1"]).ravel()
    t = np.array(f["time_vec"]).ravel()[:nt]
    case = "".join(map(chr, np.array(f["case_name"]).ravel()))
    get = lambda k: np.asarray(f["sfr3"][k][:nt, :], dtype=np.float32)   # stored as (t, x)

    groups = {
        "core":  ["eta", "ux", "uy"],                  # ~197 MB, needed for the target
        "diagA": ["dudt_T_x", "dudt_T_y", "p"],        # ~197 MB, for cross-checks
        "diagB": ["dud_x", "dud_y", "dwd_x", "dwd_y"], # ~262 MB, for cross-checks
    }
    for name, keys in groups.items():
        arrs = {k: get(k) for k in keys}
        for k, a in arrs.items():
            print(f"{name:5s} {k:9s} shape={a.shape} nan={np.isnan(a).mean():.3f} "
                  f"min={np.nanmin(a):.4g} max={np.nanmax(a):.4g}")
        np.savez(os.path.join(OUT, f"{TAG}_{name}.npz"), x=x, t=t, case=case, **arrs)

print("case:", case, "| nt =", nt, "| nx =", x.size)
```
